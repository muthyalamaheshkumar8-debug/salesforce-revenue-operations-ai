## Reconcile raw and cleaned records

```python
import pandas as pd
from pathlib import Path
ROOT = Path.cwd()
if ROOT.name == "python": ROOT = ROOT.parent
raw = pd.read_csv(ROOT / "data/raw/opportunities.csv")
cleaned = pd.read_csv(ROOT / "data/cleaned/opportunities.csv")
exceptions = pd.read_csv(ROOT / "data/cleaned/quality_exceptions.csv")
assert cleaned.opportunity_id.is_unique
assert (cleaned.amount.dropna() >= 0).all()
assert cleaned.loc[cleaned.stage.str.startswith("Closed"), "amount"].notna().all()
assert cleaned.rep_id.notna().all()
print(exceptions.groupby("disposition").size().to_string())
print(f"Retained opportunities: {len(cleaned)} / {raw.opportunity_id.nunique()}")
```
