## Validate the flat 5% portfolio simulation

```python
import pandas as pd
from pathlib import Path
ROOT = Path.cwd()
if ROOT.name == "python": ROOT = ROOT.parent
c = pd.read_csv(ROOT / "data/cleaned/compensation.csv")
assert c.rep_id.is_unique
assert (c.target > 0).all()
assert c.commission_rate.between(0,1).all()
assert ((c.actual * c.commission_rate - c.payout).abs() < 0.01).all()
print(c[["rep", "target", "actual", "achievement", "payout"]].to_string(index=False))
print(f"Total estimated payout: ${c.payout.sum():,.2f}")
```
